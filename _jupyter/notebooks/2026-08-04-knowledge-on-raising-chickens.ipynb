To prepare for raising chicken in the NE area for eggs and meat, here is the list of knowledge:

## 1. Breed ##

1.1 requirement:
+ Backyard free-range
+ Primary goal: stable egg production
+ Secondary goal: natural self-propagation in spring (hens go broody & hatch chicks)
+ Critical environment: long severe winters, temperature can drop -30℃ and below

1.2 Key cold-climate selection criteria:

+ Small / rose comb (low frostbite risk; single large combs easily freeze in winter)
+ Dense, thick feathering & compact body for heat retention
+ Moderate annual egg yield (balance between laying performance + broodiness)
+ Reliable spring broody instinct & good mothering ability
+ Tolerate rough feeding & backyard free-range conditions

1.3 Breed recommendation:
+ Lindian （林甸）
    + Origin: Local domestic breed, fully adapted to the frigid winters.
    + Egg production: 160–200 light brown eggs/year; maintains decent laying in winter with proper coop setup
    + Broodiness: Moderate. Hens naturally go broody in spring, excellent mothers, high hatch & chick survival rate
    + Traits: Strong disease resistance, great forager, calm temperament, dual-purpose (eggs main, surplus roosters for meat)

## 2. Coop & Run Design: Housing System Theory (Climate-adaptive design priority) ##

2.1 Site Selection 

+ Face the longest wall south to capture solar heat in winter; large south-facing window.
+ Build behind windbreak (house, dense shrubs) to block prevailing northwest winter winds.
+ Avoid low depressions where cold air pools and snow accumulates.
+ Elevate ground so water cannot seep under the coop.

2.2 Structure & Foundation
+ Elevated floor system to protect chickens from frost, and prevents predator tunneling. For example, Floor frame: 2×6 lumber; lay plywood, then 100 mm rigid XPS foam insulation, cover with another plywood layer. Seal all seams with foam.
   
2.3 Walls
Wall build-up outside → inside:
+ Exterior siding (plywood / metal sheet)
+ Air gap
+ Rigid XPS foam board (80–100 mm) or dense mineral wool
+ Vapour barrier (plastic sheet on warm interior side — stops condensation inside insulation)
+ Interior plywood lining

2.4 Roof (biggest heat loss point)

+ Steep pitch (≥30° slope) to shed heavy Heilongjiang snow; engineer roof for snow load.
+ Roof assembly: metal roofing / plywood → air gap → 150 mm minimum insulation → vapour barrier → ceiling plywood.
+ Install continuous ridge vent along roof peak — this is your primary permanent exhaust vent. Do NOT block it in winter.

2.3 Ventilation System — The most commonly botched part

High vents always open; low intake vents adjustable to block wind.

+ Exhaust vents: Ridge vent / gable vents, minimum 45 cm above the highest roost bar. Warm, moist air rises and exits above birds, zero direct draft on sleeping hens. Cover with heavy hardware cloth.
+ Intake vents: Lower on side walls, fitted with adjustable baffles/wind shields. In deep winter, partially close intakes to limit cold wind; open wider in milder weather.
+ Airflow goal: Coop feels cold to humans but dry. If you see frost/condensation on ceiling, you need more high vent capacity.

2.4 Roost & Nest Box Design for Cold Climate

2.4.1 Roosts

+ Height: 70–90 cm above floor (warm air accumulates higher up).
+ Use rounded solid timber (avoid metal; metal freezes).
+ Space: 30–35 cm per bird; enough room to huddle without crushing.
+ Keep roosts away from walls and any air intake to eliminate side drafts.
+ No roost directly under ventilation openings.

2.4.2 Nesting Boxes

+ Insulate all six sides of nest boxes.
+ Recessed, sheltered location away from wind.
+ Floor lined with thick straw; deep litter inside boxes.
+ 1 nest box per 3–4 hens.
+ Keep boxes darker; hens lay better and eggs stay cleaner.

2.5 Windows & Access Doors

+ Large south-facing insulated window (double-glazed if possible). Capture solar gain during short winter days.
+ Cover interior with removable clear plastic film in deepest winter to cut heat loss.
+ All doors with rubber weatherstripping; hang thick canvas thermal curtain inside doors to block wind infiltration when opened.
+ Pop door (hen exit to run) must seal tightly; install wind shield outside the pop door.

2.6 Deep Litter Method, instead of frequent full clean-out:

+ Start with 15–20 cm bedding (pine shavings / straw).
+ Add fresh dry material on top every 2–3 weeks.
+ Composting litter generates mild steady heat inside the coop and adds insulation.
+ Critical: This only works if ventilation is adequate; otherwise litter turns wet and ammonia spikes. Once yearly full clean-out in spring.

2.7 Run / Outdoor Area
   
Winter hens rarely go outside below −20°C, but you still need wind protection:

Build a 3-sided wind shelter attached to coop.
Clear snow regularly; avoid ice build-up (ice causes foot frostbite).
Consider removable wind screens (heavy tarpaulin) on north/west sides of run.

2.8 Water — Major winter challenge

Frozen water kills egg production and damages fertility. Place the coop somewhere the water won't freeze.

## 3. Year-Round Care to Maintain Fertility ##

Hens slow or pause laying in cold short-day winter conditions, and rooster fertility drops sharply in freezing weather. Follow these rules to guarantee fertile eggs the next spring:

3.1 Light management
    Chickens need 14–16 hours of daily light to maintain hormone and reproductive function. Add low-wattage barn lights on a timer in the coop during winter to hit this light requirement. This keeps both hens laying and roosters fertile all year.

3.2 Nutrition
    Feed adult flock layer feed (16–18% protein) daily. Add scratch grains, greens, oyster shell calcium, and fresh clean water nonstop. Malnutrition leads to infertile eggs, weak embryos, and thin eggshells that break during incubation.

3.3 Coop health & parasite control
    Keep bedding dry, clean droppings weekly, treat mites/lice/worms twice yearly. Stress, illness, or parasites completely ruin egg fertility and embryo survival.

3.4 Separate sick/old birds
    Cull very old hens (over 3 years old): Their egg fertility and hatch rates decline drastically. Keep only young, healthy 1–2 year old hens as your primary breeding stock.

## 4. How to Collect & Store Fertile Eggs Before Hatching ##

If you want to save eggs laid in late fall/winter to hatch early next spring, proper storage preserves embryo viability:

+ Collect eggs twice daily (morning and evening) to avoid eggs freezing overnight or sitting in hot sun, which kills the developing embryo.

+ Wipe dirty eggs gently with a dry cloth; do not wash them. Washing removes the natural protective bloom on the shell and lets bacteria seep inside, ruining the egg.
+ Store fertile eggs pointy-end down in a cool space (optimal temperature: 10–15°C / 50–59°F, 70–75% humidity).
+ Rotate eggs once per day (tilt 45 degrees side to side). This stops the yolk from sticking to the shell membrane and killing the embryo.
+ Maximum storage window: Fertile eggs stay hatchable for ~7–10 days. After 14 days of storage, hatch rates plummet severely. For next-year chicks, do not store eggs for months; instead, let the flock lay fresh fertile eggs in early spring, then immediately start incubation.

## 5. Two Ways to Hatch Fertile Eggs into New Chicks ##

You have two options once you have fresh fertile eggs in the next year’s breeding season:

5.1 Option A: Use a broody mother hen (low-effort, natural method)

Some hens go broody: they stop laying, sit constantly on a nest of eggs, and naturally incubate and hatch chicks.

+ Let a broody hen gather a clutch of 8–12 fertile eggs under her.
+ She will incubate for exactly 21 days, then care for the newly hatched chicks, keep them warm, and teach them to eat/drink.
+ Pros: No electricity or equipment needed; mother hen protects chicks from predators and weather.
+ Cons: You rely on hens naturally turning broody; some breeds rarely go broody.

5.2 Option B: Artificial incubator (controlled, reliable, scheduled hatching)

If no hens go broody, use a small home incubator:

+ Place fresh fertile eggs inside, set temperature to 37.5°C (99.5°F) and 50–55% humidity for the first 18 days.
+ Manually rotate eggs 3–5 times daily for 18 days.
+ Stop rotation on day 19, raise humidity to 65–70% for the final 3 days of lockdown until chicks pip and hatch on day 21.
+ Move hatched chicks to a heated brooder box with a heat lamp, starter chick feed, and clean water for the first 6 weeks of life.

5.3 Quick Fertility Check (Candling Eggs)

7–10 days after starting incubation, candle eggs with a bright flashlight in a dark room:

+ Fertile egg: Dark developing embryo, visible blood veins, shadowy mass inside the shell.

+ Infertile egg: Clear, yellow yolk only, no veins or growth. Discard infertile eggs to avoid rotting inside the incubator/nest.

## 6. Indoor coops ##

Two ways to interpret “indoors”

+ Attached insulated outbuilding / unheated enclosed shed (best option)

  Not inside your living house, but an enclosed space connected or near your home. This is what most northern backyard keepers choose.
  
+ Inside your residential living space (basement, spare room, garage inside the house)

  Possible, but much harder; Here are the risks:


    + Humidity & Ammonia

      Chickens produce massive moisture from breathing and manure.

      In sealed indoor rooms: humidity spikes → condensation, mould, strong ammonia gas. Ammonia irritates chicken lungs, destroys rooster fertility, causes eye damage, stops egg laying entirely. Ammonia is also bad for human respiratory health.

    + Odour

 
      Even with deep litter, indoor confinement will create noticeable smell. You need frequent bedding top-ups and strong ventilation.
    + Dust

      Chicken dander, fine bedding dust becomes airborne. Can trigger allergies.
    
    + Rooster noise

      If you keep a rooster indoors: constant crowing at dawn and throughout the day. This will become a major nuisance inside a house. Neighbours or household members may not tolerate it.
    + Biosecurity
      Chickens carry bacteria (Salmonella). You must separate their area from food prep surfaces. Do not let chickens walk freely around your kitchen/living areas.


If you plan to put them inside a spare room / basement (inside the house), the minimum design rules:

+ Fully enclosed dedicated pen
  Never free-range chickens inside your home. Build a solid barrier: plywood + hardware cloth to contain litter, dust, mess.

+ Powerful controlled ventilation (non-negotiable)

  + Small exhaust fan vented directly outdoors.
  + Do not just crack a window; in Heilongjiang winter, cold incoming air will create condensation.
  + Run fan on a timer to exchange air steadily without creating drafts on roosting birds.

+ Bedding management
  + Use deep litter system: dry straw / wood shavings. Add dry material every week. Remove wet spots immediately. Wet litter = instant smell and ammonia.

+ Lighting plan (critical for your fertile egg goal)
+ Maintain 14–15 hours total light daily with LED timer lights. Consistent light keeps hens laying and preserves rooster fertility over winter.
+ Keep light cycle regular; avoid random light at night.
+ Water：
    + Heated poultry waterer. No frozen water, no spills (spills = damp litter). Place water in a tray to catch splashes.
+ Roosts & nest boxes inside the pen：
  Same cold-climate rules: elevated roosts, insulated nest boxes away from air currents.

Insulated indoor shed / attached utility room is still strongly recommended over living room/basement.


## 7. Feeding ##

7.1 Day 1‑7: Newly hatched chicks (0‑1 week)

- Feed type: *Starter feed (20‑22% crude protein)*, fine‑ground chick starter crumbles. Do NOT use layer feed (poultry feed specially made for mature laying hens) here — too much calcium damages young kidneys.
  
- Water: Room‑temperature clean water, add a tiny amount of chick‑electrolyte for the first 2‑3 days. Waterers must be shallow so chicks cannot drown. Dip each chick’s beak into water right after placing them to teach them to drink.
  
- How to feed: Free‑choice (feed always available). Sprinkle a little feed on flat paper/towel for first 3 days so chicks can easily find food; then switch to chick feeders.

  
- No treats yet: No grains, no kitchen scraps for week one. Their digestive system is too immature.

  
- Environment note: Keep brooder temperature 32‑34°C the first week. Cold chicks stop eating and die quickly.

7.2 Week 2‑6: Young chicks (2‑6 weeks old)

- Feed: Keep using **chick starter (20% protein)**. You can move from fine crumble to slightly larger crumble. Still free‑access feeding.
  
- Water: Fresh clean water daily. Remove bedding debris from water.

  
- Small treats (after week 3 only): Very small amounts of finely chopped greens, hard‑boiled egg yolk. Treats ≤5% of total daily food. Too many treats displace balanced feed and cause stunted growth.

  
- Grit: If you give treats/greens outside formulated feed, provide fine‑chick grit (insoluble granite). Chicks have no teeth; grit grinds food in gizzard. Do not mix shell grit at this stage.

7.3 Week 7‑18: Grower / Pullet phase (before laying starts)

- Feed type: Switch to **grower / pullet feed (16‑18% protein)**. This is lower‑calcium feed. *Still NO layer feed*.
  
- Feeding style: Continue free‑choice feeding. Do not restrict food for backyard home‑flock.

  
- Treats: Can add limited whole grains, vegetable scraps, weeds. Treats still under 10% of daily intake.

  
- Grit: Supply medium‑size poultry grit free‑choice.

  
- Important: If you have roosters growing alongside pullets, everyone eats the same grower feed at this stage. Roosters do not need extra calcium.

- Note on sexual maturity: Most heritage breeds start laying around week 18‑22; slower cold‑hardy breeds may go to week 24‑26. Watch for first egg signal: redder combs & wattles.

7.4 Point‑of‑lay transition (1‑2 weeks before first egg appears)

Slowly transition to layer feed. Do NOT abruptly switch.

- Around week 17‑19: Gradually mix layer feed into pullet feed over 7‑10 days.
- **Layer feed specs**: 15‑16% protein, high calcium (3.5‑4%). Calcium builds strong eggshells.


7.5 Adult laying hens & roosters (From first egg onward)

7.5.1 Hens (laying)

- Main diet: **Layer feed, free‑choice**. Always keep feed available.
- Calcium backup: Offer separate oyster‑shell / crushed‑limestone in a separate small dish. Hens self‑regulate calcium intake. This is especially important in cold winters and for fertilized‑egg production.
- Treats limit: Treats, scratch grain, kitchen scraps should make up **max 10 % of total daily food**. If treats exceed this, egg production drops, shells weaken, body condition suffers.
- Safe treats: leafy greens, cabbage, carrots, leftover cooked rice/grains, mealworms.
- Bad treats to avoid: salty food, moldy food, raw potato peels, chocolate, onions in large quantity, avocado skin/pit.

7.5.2 Roosters housed with laying hens

Roosters cannot eat only layer feed long‑term because high calcium harms male kidneys.
Two practical solutions for your mixed flock:

+ **Best method**: Keep pullet/grower feed available in a feeder only roosters can reach (higher height hens cannot access), plus layer feed for hens.
+ Simpler backyard workaround: Limit layer feed to 90% of diet, supplement whole grain, and provide separate oyster‑shell dish so hens self‑take calcium. Do not feed pure layer feed exclusively long‑term to a flock with roosters.

7.6 Special feeding rules for extremely cold winters

Cold weather increases energy demand. Chickens burn extra calories to stay warm.

+ Keep complete formulated feed as primary food. Do not over‑load on scratch grain only; grain is mostly energy but lacks vitamins/minerals for egg‑making and fertilized‑egg quality.
+ You can give small whole grain scratch in late afternoon. It generates extra internal heat overnight during cold nights.
+ Water is critical: Water must stay liquid. Frozen water = chickens stop eating, egg‑laying stops. Use heated poultry waterer.
+ Protein intake should not drop in winter. Low protein = fewer fertilized eggs, poorer hatch rates when you want to breed new chicks.
+ Continue free‑choice oyster shell for laying hens in winter. Cold stress makes eggshell quality worse.

7.7 If you want fertile eggs for hatching (your goal for next‑year new chicks)

Nutrition directly impacts rooster fertility and egg hatchability:

- Roosters need adequate protein, vitamin A, vitamin E. Don’t starve roosters; don’t over‑treat with grain.
- Hens in breeding condition need good calcium, phosphorus. Poor nutrition leads to fertile eggs that fail to hatch.
- Avoid feeding heavily mold‑contaminated grains; mold toxins kill developing embryos inside fertile eggs.
