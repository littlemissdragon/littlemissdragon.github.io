## Introduction ##

This is a structured execution plan for relocation to a remote city. It has pre-arrival, arrival-day, day-2, day-3 workflows for rental search, tiered contingency scenarios, budget guardrails, and handoff to the later home-buying phase.

**Core Objective**: Secure an affordable, reliable long-term rental within **2–3 calendar days after arrival**, minimising cash burn from hotel accommodation, then launch a home-purchase search from a stable rental base.

**Key Constraint**: Low inventory on mainstream real estate websites; most private rentals are off-platform and found via local social media, word-of-mouth and local agents.

**Risk Focus**: Hotel cash bleed, fake listings, slow landlord response, no available unit on arrival, property issues (leaking, heating, plumbing).

## 1. Pre‑Arrival Phase (7–10 days before departure)

Goal: Warm up local leads before you arrive, so you do not start from zero when you check into the hotel.

### 1.1 Local Market Recon

+ Map 3–4 **target neighbourhoods** first with the following priorities:
  
    + proximity to the zones you will later shop for a purchased house
    + with basic furnishing (bed, eletricity, water) provided to lower the overall cost of renting
    + consider house first, then apartment
    + grocery access

+ Collect baseline data:

    + local average rent
    + standard lease terms
    + typical deposit structure
    + who pays heating/electricity/water/internet

+ Build a shortlist of 2–3 local independent **real estate agents** who specialise in private off-market rentals. Prepare the requirements in advance, and get ready to contact them if you fail to find rental properties on your own within 2 days. 

### 1.2 Social Media Rental Advertisement Draft & Group Prep

Prepare your ad ready to post the moment before you arrive.

**Ad template (can be translated for local groups)**:

> 
> Looking for a private apartment / house for long-term rent, [city name]. Move-in ASAP. Budget: [X]. Need reliable heating, intact plumbing, quiet neighbourhood. I am a serious, stable tenant, no pets / clean tenant profile [adjust]. Can view in person within 24 hours after contact. Direct landlord preferred; no agency scams. Please send location, price, photos and heating condition. DM / call: [contact].

Action steps:

- Join local WeChat groups, neighbourhood groups, local life / second-hand / housing groups ahead of time.
- Set a posting cadence: post on arrival, repost lightly once every 12 hours, avoid spamming.
- Prepare a standard screening questionnaire to send to every respondent: monthly rent, deposit, heating included, lease length, whether owner lives on-site, maintenance responsibility, can you register address, any restrictions on visitors / work.

### 1.3 Hotel Booking Rules (Cash Burn Control)

- Book on a **flexible, cancellable rate**, with daily or 3-day blocks. Do not pre-pay a long hotel stay.
- Select an extended-stay hotel / family hotel with kitchenette and laundry if possible; this reduces extra eating-out costs.
- Hard budget cap: define a maximum acceptable hotel spend per day, and a **hard stop threshold**: if rental leads are zero after 48 hours, trigger the contingency housing tier.
- Keep a separate “hotel burn tracker”: log daily cost, remaining budget buffer, and how many viewing slots you have scheduled.

### 1.4 Document & Deposit Prep

Have digital copies ready: ID, proof of stable income, bank statement.
Prepare a fixed rental deposit fund ready for transfer **only after in-person inspection and signed lease**.
Rule: **No deposit sent before physical viewing**. This is your main anti-scam rule.

## 2. Arrival Day (Day 1: Check-in + Active Lead Generation)

Goal: Confirm hotel base, launch ads, lock 3–4 viewing slots for Day 2.

1. Check in, confirm hotel cancellation policy, confirm checkout time.
2. Post your rental advertisement in all pre-joined local groups.
3. Follow up with pre-contacted local agents, tell them you are in the city and ready to view immediately.
4. Screen all incoming messages using your questionnaire; filter out:
   - Unwilling to share exact address
   - Landlord refusing in-person meeting
   - Price far below market (high scam risk)
5. Schedule viewings grouped by geographic zone to avoid wasted travel time.
6. Evening review: tally qualified leads, mark high-priority units, confirm next-day viewing sequence.

## 3. Day 2: Viewing, Verification & Negotiation

Goal: Inspect units, validate heating, plumbing, insulation, lease terms; aim to sign if one unit meets criteria.
Inspection checklist for every viewing (critical for cold northeast climate):

- Heating system: is it working, who pays heating bills, how reliable is supply in deep winter
- Water pipes: insulation, risk of freezing
- Window sealing, draft, roof / wall moisture
- Electrical safety, wiring, breaker condition
- Lock security, privacy, noise level
- Furniture / appliance inventory, photo every defect before signing
- Lease clauses: notice period, repair obligations, deposit return rules, rent increase terms

Negotiation priority: flexible move-in date, clear heating responsibility, written deposit return terms.
If you find a qualifying unit: sign the written lease, take photos of all pages, pay deposit, confirm move-in time ideally same-day or early Day 3.

If no qualifying unit on Day 2: activate secondary channels: ask hotel front desk, shop owners, grocery staff for private landlord referrals. Local word-of-mouth often finds unlisted homes.
