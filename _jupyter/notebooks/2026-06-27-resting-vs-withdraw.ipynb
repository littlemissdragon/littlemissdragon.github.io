```python
# Import libraries
import matplotlib.pyplot as plt
import numpy as np

# Set plot style
plt.rcParams["font.family"] = "sans-serif"
plt.rcParams["figure.figsize"] = (16, 10)
plt.rcParams["font.size"] = 11

# --------------------------
# Core comparison data (1:1 from your article)
# --------------------------
categories = [
    "Core Definition",
    "Root Cause",
    "Inner Motivation",
    "Key Behavioral \nSigns",
    "Internal Feeling",
    "Short & Long-Term \nConsequences",
    "Core Purpose",
    "Post-Downtime \nState"
]

rest_data = [
    "Intentional, peaceful pause after \nconsistent effort for physical/mental recovery",
    "Genuine physical & mental fatigue \nfrom long-term work output",
    "I have worked enough; I pause to \nrecharge for future productivity",
    "Conscious pause, calm mood, easy \nto resume work, no guilt/numbness",
    "Relaxed, stable, peaceful, no fear \nor shame",
    "Restores energy, improves focus, \nsustains long-term productivity",
    "Heal and replenish myself",
    "Refreshed, motivated, grounded"
]

withdrawal_data = [
    "Subconscious emotional avoidance; \nescape from fear, shame and anxiety; \n detachment from reality",
    "Conflict between learned self-worth \njudgement and innate human resistance \nto feeling worthless",
    "Feeling inadequate to make any progress, \ntrying risks failure and feeling worthless",
    "Numb distraction binge, avoid work/\nresponsibilities, self-justify inactivity",
    "Empty, sad, anxious, restless, \nurge to run away from reality",
    "Worsens rationality and productivity, \ndrops self-esteem, deepens anxiety cycle",
    "Hide from emotional pain, responsibility, and \nreality",
    "Numb, guilty, stuck, more anxious \nand unmotivated"
]

# Combine data for table
table_data = np.column_stack([categories, rest_data, withdrawal_data])
col_labels = ["Category", "Rest", "Withdrawal"]

# --------------------------
# Create table plot
# --------------------------
fig, ax = plt.subplots()
ax.axis("tight")
ax.axis("off")

# Create main table
table = ax.table(
    cellText=table_data,
    colLabels=col_labels,
    cellLoc="left",
    loc="center",
    colWidths=[0.15, 0.4, 0.4]
)

# Table style settings
table.auto_set_font_size(False)
table.set_fontsize(16)
table.scale(1, 5)

# Header color styling
for i in range(3):
    table[(0, i)].set_facecolor("#2c3e50")
    table[(0, i)].set_text_props(color="white", weight="bold")

# Alternate row background color for readability
for row in range(1, len(categories)+1):
    color = "#f8f9fa" if row % 2 == 0 else "white"
    for col in range(3):
        table[(row, col)].set_facecolor(color)

# Add title
plt.title(
    "True Rest vs. Emotional Withdrawal (Seasonal Anxiety Pattern)",
    fontsize=14,
    pad=20,
    weight="bold"
)

plt.tight_layout()
plt.show()
```

## Why withdrawing happen ##

Doing nothing but thinking is pure strategy, which only lead to depth temperarily, not breadth. And eventually depth development will stop as well, because it depends on the development of breadth.

## Withdrawing and trauma ##

When facing a problem, if you withdraw, you enter a frozen state. If you cannot come out of that state, you become traumatized.  Next time when you face the same problem again, you are more likely to enter that frozen state. The problem then becomes a trigger for feeling powerless.
