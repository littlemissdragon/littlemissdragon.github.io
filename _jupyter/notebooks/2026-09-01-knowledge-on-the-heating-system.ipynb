## 1. Introduction ##

The far northeastern area features **extremely cold winters**: outdoor temperatures can drop to ‑30℃ ~ ‑35℃, long heating season (about 6‑7 months per year), large temperature difference between indoor and outdoor, high risk of pipe freezing. All systems must prioritize **anti‑freezing performance, low‑temperature adaptability, safety, operating cost, and backup solutions**.

## 2. Options ##

2.1. Solid‑fuel boiler (water circulation)

- Core principle: Hot‑water boiler burns solid fuel; hot water circulates to radiators / underfloor heating
- Low‑temperature adaptability: ★★★★★, works reliably down to −35 °C
- Installation complexity: Medium‑high; requires dedicated boiler room, chimney, full water circuit
- Running cost: Medium‑low; biomass pellets cleaner than raw coal
- Safety notes: Fire & carbon‑monoxide hazards; reliable chimney and ventilation are mandatory
- Efficiency & real‑world performance: Very stable heat output; maintains 18‑22 °C indoor; requires manual fuel feeding and ash removal

2.2. Low‑temperature rated air‑source heat pump

- Core principle: Extract heat from outdoor air via compression cycle for water‑loop heating
- Low‑temperature adaptability: ★★★☆☆; **must select −35 °C cold‑climate model**. Standard units fail below −20 °C
- Installation complexity: Medium; outdoor unit + indoor water circuit / fan coils
- Running cost: High power consumption in deep winter; COP drops sharply below −25 °C
- Safety notes: Electrical safety; pipeline antifreeze; add propylene‑glycol antifreeze solution
- Efficiency & real‑world performance: Good performance above −20 °C; big output drop in mid‑winter; auxiliary electric backup heat is usually needed

2.3. Ground‑source (geothermal) heat pump

- Core principle: Heat exchange with constant‑temperature underground soil
- Low‑temperature adaptability: ★★★★★; performance independent of outdoor air temperature
- Installation complexity: Very high; vertical borehole drilling required, needs large land area
- Running cost: Low long‑term operating cost; very high upfront capital investment
- Safety notes: Low risk; closed underground fluid loop
- Efficiency & real‑world performance: Stable consistent heating all winter; poor choice for sites with limited land

2.4. Electric heating (electric boiler / electric floor heating / infrared radiation heaters)

- Core principle: Direct electricity‑to‑heat conversion
- Low‑temperature adaptability: ★★★★★; no exposed outdoor pipes
- Installation complexity: Low‑medium
- Running cost: **Very high** for 6‑7‑month continuous winter operation
- Safety notes: Risk of overheating and electric leakage; high‑quality ground‑fault protection required
- Efficiency & real‑world performance: Simple setup; best used as backup / auxiliary heat source rather than main heating

2.5. Traditional masonry stove‑kang (wood / biomass burning, plus radiator loop)

- Core principle: Burn firewood / biomass to heat brick kang bed plus connected radiators
- Low‑temperature adaptability: ★★★★★
- Installation complexity: Medium; requires on‑site masonry work
- Running cost: Very low if local firewood / biomass is available
- Safety notes: High carbon‑monoxide risk; smoke‑exhaust chimney design is critical
- Efficiency & real‑world performance: Good for living quarters; difficult precise temperature control; poor for multiple separated rooms


## 3. Key pre‑design parameters you must confirm before purchase & installation ##

These decide whether your heating works reliably in‑30℃ weather.

3.1. Heat load calculation (most critical step)

Rule: Do not buy heating equipment only by room area; calculate **total heat loss**. 

- Severe cold northeast area: typical heat demand: **80‑120 W per square meter** for well‑insulated buildings; 120‑180 W/m² for poorly insulated old buildings.
  
- Chicken coops: add extra heat loss from ventilation air exchange. Poor insulation will double your heating cost, no matter which heating device you buy.

3.2. Building insulation status

- Wall insulation, roof insulation, double/triple‑layer sealed windows are the foundation of heating. Without insulation, even the largest heating system cannot keep warm and consumes huge amounts of energy.
  
- For chicken houses: wall + ceiling insulation is non‑negotiable priority.

  
3.3. Local resource availability

- Is cheap biomass pellet / firewood accessible? (for solid fuel systems)

- Is your transformer capacity enough for high‑power electric / heat‑pump equipment? Rural northeast grid sometimes has power‑capacity limits.

- Land space: can you place boiler room, outdoor heat‑pump unit, or drill boreholes for ground‑source heat pump?

  
3.4. Target indoor temperature

   - Human living space: 18‑22℃
     
   - Winter laying‑hen chicken house: typical target 12‑18℃ (avoid over‑heating to save energy)

## 4. System setup & installation core requirements (water‑circulation systems: most widely used in northeast)

Most practical systems use **hot‑water circulation**: 
+ heat source →
+ circulating pump →
+ pipeline →
+ terminal heat emitter (radiator / floor heating) →
+ back to heat source.

4.1 Mandatory anti‑freezing rules for far‑northeast

+ Any water‑filled pipe exposed outdoors or in unheated space **must add food‑grade propylene glycol antifreeze**. Pure water will freeze, expand and crack pipes at sub‑zero temperature.
  
+ Circulation pump:

    + electrical pump: must have **anti‑freeze auto‑run function**. If power fails for long hours, you need backup power generator, or drain all water from pipes. Power failure is a real risk in rural northeast winter.
    + No pump method: Thermosiphon (thermal siphon circulation, most common traditional no‑pump method)

        - Works on the principle: warm water is less dense and rises; cold denser water sinks.
        
        - No electricity, no moving mechanical parts.
        
        - Setup requirement: heat source must sit **lower** than the storage tank. Heated water naturally flows upward into the tank; cooler water from the tank drops back down to the heater, creating continuous loop circulation.
        
        - Common use: old‑style solar water heaters, antique wood‑fired boiler domestic hot‑water and simple heating loops.
        
        - Limitations: depends on consistent temperature difference; flow rate is gentle; pipe layout and vertical height difference are critical; performance drops in very cold conditions.
          
+ Pipeline layout: bury outdoor pipelines below local frost depth (Heilongjiang **frost depth reaches 1.8‑2.2 meters**). If exposed above ground, use thick multilayer thermal insulation shell plus waterproof outer wrapping.

4.2 Terminal heat emitters choices

+ **Cast‑iron radiators**: classic northeast choice. High heat capacity, keeps radiating heat even when circulation stops temporarily. Easy maintenance. Good for both house and chicken barn.
  
+ **Water underfloor heating**: gentle uniform heat. Slow thermal response; needs careful antifreeze protection. Not ideal if frequent power outages happen.

  
+ Fan‑coil units: fast temperature rise, but produce dry airflow; not recommended for chicken housing.

## Safety checklist for cold‑climate heating ##

+ **Carbon monoxide prevention (solid‑fuel burning systems)**
   - Boiler / stove must have independent, properly height‑designed chimney, no leakage smoke joints.
   - Install CO alarm detector in living areas. Never seal the boiler room completely tight; keep proper fresh‑air intake for combustion.
     
+ **Electrical safety (heat‑pump / electric boiler)**
   - Ground‑fault circuit breaker for all high‑power heating equipment.
   - Outdoor unit wiring needs cold‑resistant cable, moisture‑proof junction boxes.
+ **Pressure safety for water heating loop**
   - Equip expansion tank, pressure relief valve. Prevent pipe rupture caused by water thermal expansion.

+ Power‑failure contingency

In deep‑winter power outage: circulating pump stops. Water in pipes risks freezing.
   - Options: small gasoline generator for critical heating‑system components; or ability to fully drain the whole water circuit.

## Mechanism of Hydronic (Water Circulation) Heating System ##

For your 35–40 m² small house + mixed gravity/electric pump solid-fuel boiler setup:

1. Where water is heated

The heat source is your **solid-fuel boiler** (wood/coal pellet stove boiler).

- Inside the boiler is a water jacket / heat exchanger: fire and hot flue gases surround this metal chamber.
- Cold water flows into this chamber, absorbs combustion heat, and becomes hot water (typically 60–85°C for radiators).

2. Two circulation modes (matches your power-outage hybrid design)

2.1 Mode A: Electric pump circulation (normal operation)

+ The boiler heats water.
+ A small circulator pump pushes hot water out from the boiler’s top outlet.
+ Hot water travels through supply pipes to wall-mounted radiators in rooms.
+ Radiators release heat to the air; water cools down.
+ Cooled return water flows back via return pipes into the bottom inlet of the boiler to be reheated.

This is fast, allows smaller pipe diameters, and works well for compact homes.

2.2 Mode B: Gravity / thermosiphon circulation (power outage backup, NO pump)

This is the passive backup you want for heavy snow power cuts.
**Core physical principle: hot water is less dense and rises; cold denser water sinks.**

+ Water heated in the boiler expands, density drops → hot water naturally rises upward out of the boiler.
+ Piping must be sloped correctly, with large enough pipe size, and boiler installed **lower than your radiators**.
+ Hot water rises to radiators, releases heat, cools.
+ Cooler heavier water sinks down the return pipe back into the boiler.

No electricity required. But flow rate is slower, heat output lower, pipe layout is much more sensitive. Gravity flow needs a vertical height difference between boiler and radiators, plus larger pipe sizes. Air lock is the biggest failure risk for passive mode.

3. Radiator heat transfer

Hot water flows through radiator internal passages. The metal radiator conducts heat, then transfers heat to room air by convection + minor radiation. Air in the room circulates naturally around the radiator.

4. Auxiliary components (basic system)

- Expansion tank: absorbs volume expansion when water heats up (critical for closed systems; prevents pipe/boiler bursting)
- Air vents: bleed trapped air (air stops water flow, especially bad for gravity loop)
- Valves: isolate boiler/radiators for maintenance
- Pressure gauge (closed system)

5. Hybrid system logic (your design)

- Power available: electric pump runs, full heating power.
- Power failure: shut pump off, the system automatically switches to thermosiphon gravity circulation, limited but usable heat from the solid fuel boiler.


## Thermosiphon Mechanism ##

1. Step-by-step flow

+ Water in the boiler absorbs heat from burning wood/coal, warms up, becomes lighter.
+ Hot water rises naturally upward through the **riser pipe** to your radiators on the upper level of the house.
+ Hot water releases heat in radiators, warms the room. After cooling, the water becomes denser and heavier.
+ Cold dense water flows down the **return pipe** back into the bottom of the boiler.
+ The cycle repeats continuously as long as the boiler keeps heating water.

2. Critical design rules (very important for your 35–40m² cabin in cold snow areas)

- **Boiler must sit lower than radiators**. The vertical height difference creates the driving pressure. More vertical drop = stronger circulation.
- Pipe size: Use larger diameter pipes than forced pump systems (small pipes create too much friction and stop flow).
- Gentle upward slope for hot supply pipe; continuous downward slope for return pipe, no low pockets where air can trap.
- Air vent at the highest point of the piping. Trapped air blocks thermosiphon flow completely.
- The whole system is a closed loop, filled with water.

## General radiator rules ##

- All radiators mounted **higher than boiler outlet**
- All radiators at similar vertical elevation (avoid one radiator far higher than others)
- Mount on exterior walls, ideally under windows
- Bottom of radiator: 8–12 cm clearance above floor
- Gap 3–5 cm between radiator and wall
- No full-length curtains, furniture or cabinets blocking front/top
- Each radiator fitted with: manual balancing valve + air bleed valve (top corner)
- Reflective aluminium foil behind radiators on outside walls
  
## Dual-mode pipe layout (pump primary + thermosiphon backup, 35–40m² home) ##

Design priority: satisfy thermosiphon rules so heating still works in power outage; pump handles normal daily operation.

- Boiler sits at the **lowest point** of the entire water circuit.
- Hot supply pipe: rises gently (slope ~1–2%) from boiler outlet up toward radiators.
- Return pipe: slopes steadily down (1–2%) from radiators back to boiler inlet.
- Keep horizontal pipe runs short; minimize long flat segments. Flat pipe kills gravity flow.
- All radiators must be mounted higher than the boiler’s water outlet.
- Keep radiators at roughly similar vertical height. One much-higher radiator will dominate flow in thermosiphon mode.
- Install manual balancing valves on every radiator:
  - Adjust for even heat when pump runs.
  - Leave valves fully open for passive thermosiphon operation during blackouts.
- Fit air bleed valves at the highest point of each radiator and the overall piping high point. Air locks stop both pump and gravity circulation.
- Use larger pipe diameter than a pump-only system: 1¼” or 1½” for main lines. Wider pipes reduce friction for slow thermosiphon flow; pump works fine with bigger pipes.
- Minimize sharp 90° elbows. Use sweeping bends to cut flow resistance (critical for gravity, also helps pump efficiency).
- Locate pump on the **return line** (cooler water side, less stress on pump). Add isolation ball valves on both sides of the pump.
- Install a bypass / isolation setup for the pump: when power fails, you can close pump isolation valves so water flows through the gravity loop, bypassing the pump.
- Include an expansion tank mounted on the hot supply side, above the boiler, to absorb thermal expansion.
- Avoid low pockets in pipes where water can trap air or sediment.
