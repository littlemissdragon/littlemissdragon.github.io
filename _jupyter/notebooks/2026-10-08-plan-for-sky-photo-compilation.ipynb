## Define the Goals (Exits) of the Project ##

+ Success exit: finish the video and upload it to YouTube. 

+ Pause exit:
    + Collect the photos into one album
    + Compile the photos into one video
    + Add background music and title
    + Output video and upload it

+ Scale‑down/abandon exit: reduce the sessions or abandon for now

## Step 1: Time Budget ##

1. Session 1 (max 45 min): Sort & pick sky photos. 
2. Session 2 (max 45 min): Import into editor, arrange sequence, set base timing.
3. Session 3 (max 45 min): Add music, title, render draft.
4. Session 4 (max 45 min): Output video and YouTube title + description + upload.

Work on the project on Tuesdays and Saturdays, one session por day maximum. 

## Step 2: Three Concrete Exit Triggers (decision‑points)

At the end of every work‑session, ask:

> *If this project was handed to me right now, exactly in its current state, would I volunteer to spend the next planned block of time on it?*

+ Case A → Yes: Proceed to next session.
+ Case B → No. You get 3 choices, no shame:

    1. Scale‑down exit (preferred). Cut features. Example: skip music and title. Finish a minimal‑viable version and upload that.
       
    2. Pause‑exit: Freeze the project. Come back another day; do not force yourself to grind today.
    
    
    3. Abandon‑exit: Accept you will not finish this. You can revisit the idea months later if interest returns.


## Step 3: Define Minimum Viable Product ##

MVP:

If these items are checked, the project counts as complete:

+ Selected set of sky photos sequenced one after another
+ Background music and Title added
+ Rendered to video file
+ Uploaded to YouTube

## Step 4: Wishlist

Whenever a new idea pops into your head mid‑work, write that new wish onto a “Wishlist”. Only touch wishlist items if:

‑ All minimum‑done requirements are already finished
‑ You still have remaining time‑budget
‑ You still feel motivated.

If any of those three is false, wishlist stays for another day, or forever.
