## Introduction ##

To understand how to make money, we first need to study the mechanics of our modern monetary system - how is money created. 

In this article, we will explore the debt-money system.

## Money as Debt ##

In a modern economy, most money (typically over 90%) is not physical cash printed by the government. Instead, it is bank deposit money created electronically by commercial banks when they make loans.

Here's the magic trick:

+ You go to a bank for a $10 loan.

+ The bank doesn't take $10 from someone else's pile of cash.

+ Instead, it creates a new $10 deposit in your account by typing numbers into a computer. This new deposit is new money.

+ In return, you give the bank an IOU (the loan contract) for $10 + 1 (we assume the interest rate is 10%).

## The Mathematical Flaw of the Debt-Money System ##

Now, let's imagine a simplified, closed economy to see the problem. Assume:

+ There is only one bank.

+ All money in existence was created by this bank as loans.

+ The total principal of all outstanding loans is $100.

+ The interest rate is 10%, so the total amount to be repaid is $110.

The critical question is: **Where does the extra $10 for the interest come from?**

+ The bank created and lent out only $100 (the principal). It did not create the extra $10 needed to pay the interest.

+ This means that the total amount of money in circulation ($100) is less than the total debt owed to the bank ($110).

It becomes a mathematical impossibility for everyone to repay their loans in full at the same time. The system is structurally designed for a shortfall.

## The Consequences: How the System Stays Afloat ##

If it's mathematically impossible, why doesn't the entire system collapse immediately? It's because the system is dynamic, not static. It avoids collapse through several mechanisms:

A. Continuous New Lending

This is the primary solution. For the extra $10 in interest to be paid, the bank must create and lend out new money. This new money provides the funds for borrowers in the old loans to pay their interest.

+ In our example, the bank must issue at least $10 in new loans just to provide the money to pay the interest on the old ones.

+ This creates a growth imperative: the money supply (and therefore debt) must **constantly expand** to service the existing debt. It's like a treadmill that keeps speeding up.

B. Defaults and Bankruptcies

The system doesn't require everyone to pay back their debt. Some borrowers will fail. Defaults and bankruptcies destroy money (the loan asset is written off the bank's books) and effectively write off the impossible-to-repay debts. This is a painful but necessary pressure valve for the system.

## Debt - The Systematic Motive to Make Money ##

The Debt-money system is structurally designed to place the entire population in a state of net debt, forcing them to compete in a game where, in the aggregate, it is mathematically impossible for everyone to win.

It systematically indebts the population and forces them into a relentless, competitive struggle for a money supply that is perpetually insufficient to extinguish the debts it represents. It is a system that requires most people to lose (through default, stress, or perpetual debt servitude) for the system itself to continue functioning for the few at the top.

The Inescapable Mathematical Reality

    The Money Supply = Total Principal: All money is created as a loan principal. If the total money supply is $X, then the total principal debt owed to banks is also $X.

    The Repayment Amount = Principal + Interest: The total amount that must be paid back to the banking system is $X + $Y (where $Y is the total interest).

    The Fatal Flaw: The money to pay $Y was never created. It does not exist in the economy.

This creates a simple, inescapable equation:
Total Debt > Total Money

The Systemic Consequences: A "Rigged Game"

This mathematical flaw doesn't mean every individual is in debt. It means that the economy as a whole is a debtor to the banking system. The "game" of the economy is not about creating value to earn a neutral currency; it's about competing to capture the limited supply of money that exists in order to service the exponentially growing debt that the money represents.

Here’s what this looks like in practice:

    A Zero-Sum Scramble for Survival: Since the money to pay all debts (principal + interest) doesn't exist, the economy becomes a fierce competition where participants must:

        Take on new debt (providing the money for others to pay their interest, but deepening the overall problem).

        Outcompete others to capture their money (by selling them goods/services, beating them for jobs, etc.).

        Force others into default. For you to be solvent and debt-free, someone else, somewhere in the system, must fail and default on their loan. Defaults are not a bug; they are a necessary feature for the system to "balance its books."

    The Growth Imperative as a Treadmill: The only way to temporarily avoid a system-wide collapse is through perpetual economic growth, which requires perpetual new debt creation. We must constantly "borrow from the future" to pay the bills of the present. This is why our economic model is so obsessed with GDP growth at all costs—stagnation means immediate crisis.

    The Redistribution Mechanism to the Financial Sector: Interest payments act as a continuous, systemic siphon, transferring real wealth (houses, labor, goods) from the broader population to the holders of debt (the financial system). You work for years to pay your mortgage; the bank created the principal with a keystroke and collects the interest as pure profit.

The Perfect Analogy: A Poker Game

Imagine a poker game where the "house" (the bank) doesn't just take a rake, but provides all the chips as loans at 10% interest.

    The house lends $100 worth of chips to the players.

    At the end of the night, the players collectively owe the house $110.

    But there are only $100 worth of chips on the table.

How can the players possibly pay back the house?
They can't. The only way is for the house to lend them more chips the next night, so the game can continue and the players can win enough new chips to pay the interest on the old chips. The game can never end, and the players are permanently indebted to the house, forced to play every night just to stay afloat.

This is exactly how our economy works. We are all the players, and the debt-based monetary system is the house.
