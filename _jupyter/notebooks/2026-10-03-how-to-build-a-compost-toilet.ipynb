## How a compost toilet works ##

A compost toilet separates **solid waste (feces)** and **urine** (most simple designs do urine diversion, this is critical to avoid stench and speed composting).

+ **Urine diversion**: Urine flows into a separate container. Urine is high in nitrogen, and if mixed with solids it creates smelly, soggy sludge. Diluted urine can later be used as garden fertilizer (1:10 water dilution, don’t use on food crops directly).
  
+ **Solid waste + carbon cover**: After pooping, you add a dry carbon material: sawdust, wood shavings, dry leaves, straw. This balances nitrogen, absorbs moisture, feeds compost microbes, and suppresses odor.

  
+ **Composting process**: Aerobic bacteria and fungi break down human waste. The pile needs **air, moisture, carbon, and time**.
   - Hot compost (130–160°F / 55–70°C): kills pathogens, takes ~3–6 months.
   - Cold compost: slower, 12+ months; must never use the end compost on raw-eaten vegetables/fruit.

+ **Finished product**: After full curing, it becomes humus-like compost. **Important safety note**: local laws vary widely. Many regions restrict humanure use, and it should NOT touch edible root crops or food you eat raw.

Core principle: **keep solids dry, keep air flowing, separate urine, give time to decompose.** It is NOT a septic tank. No water flushing required.

## Finished Product Application rules ##

- Never spread fresh compost-toilet material directly on growing plants.
- Apply to soil **before planting**, mix well into soil.
- Keep finished compost off edible plant parts:
  - Root vegetables (carrots, potatoes, beets): extra careful. Many guidelines recommend using this compost only for fruit trees, berry bushes, or non-edible ornamentals if you want zero risk.
  - Above-ground veggies (tomato, cucumber): safer, still avoid splashing compost on fruit/leaves.
- Follow a waiting period: **at least 120 days between compost incorporation and harvest** (standard food safety guideline for humanure).

3. **What to avoid adding to the compost toilet**
No bleach, harsh cleaners, antibacterial soap, medication waste, heavy metals. These survive composting and end up in your food. Use only natural carbon cover (sawdust, dry leaves). Urine is high nitrogen; if mixed, it can raise salt levels, so proper dilution is needed if used separately.

## Risk note

Even with good process, some home DIY systems cannot reliably sustain consistent hot compost. Many organic certification standards **do NOT allow humanure for certified organic food crops**. If you sell produce, you cannot use it. For personal home consumption, strict long curing + hot compost reduces risk.

## DIY Urine-Diverting Compost Toilet (Backyard, low cost) ##

1. Safety & legal pre-check first

    + Site rules: place **at least 100ft / 30m away from wells, streams, ponds**. On high, well-drained ground, away from flood zones.
    + This design is for outdoor yard use only, not inside a house.
    + Always wash hands thoroughly after use.

2. Materials list

    - 2 x 5-gallon food grade plastic buckets (for solids; 1 for use, 1 curing)
    - 1 smaller bucket (urine collection, ~2–3 gallon)
    - Toilet seat + lid (standard)
    - Plywood (1/2 inch or thicker) to build the box frame
    - 2x4 lumber for frame support
    - Silicon caulk, screws
    - PVC pipe + fittings (½" or 1") for urine diversion channel
    - Carbon cover: sawdust / wood shavings (best), dry leaf mold, straw
    - Optional: small vent pipe (2" PVC) up through the box top to reduce odors
    - Optional: screen for vent end to keep bugs out

3. Step-by-step build procedure

    3.1 Step1: Build the wooden box stand

    + Build a sturdy rectangular frame from 2x4s. Dimensions to fit your two buckets side-by-side: ~32in long × 20in deep × 18in tall (adjust for your bucket size).
    + Cut a plywood top plate. Cut **two holes**:
       - Larger hole: fits the solids bucket, sized so bucket sits flush under the toilet seat. Mount the toilet seat over this hole.
       - Smaller hole beside it: for the urine diversion trough leading to the urine bucket.
    + Attach plywood top to the 2×4 frame with screws. Seal plywood edges to resist rot (paint or linseed oil).

    3.2 Step2: Make urine diversion separator

    + Shape a smooth sloped trough (can use food-safe plastic sheet or molded PVC) under the front part of the toilet seat opening.
    + Trough slopes forward into the small urine bucket. The back part of the seat opening drops solids straight down into the solids bucket.
    + Seal all gaps with silicone to stop leaks.
    + Optional vent: drill a hole in the rear of the box, run 2" PVC pipe up 2–3ft above the roofline of your box. This pulls air out and eliminates almost all smell. Add insect screen on the pipe outlet.

    3.3 Step3: Set buckets

    + Put one solids bucket under the seat hole (active bucket). Second solids bucket set aside for curing.
    + Place urine bucket under the urine trough. Lid for urine bucket to reduce ammonia smell.
    + Keep a small bucket of sawdust + scoop next to the toilet.

## Usage procedure ##

1. Sit normally. Urine flows to urine bucket; solids drop to solids bucket.
2. After solids: add **1–2 scoops of dry sawdust/shavings** to cover completely. This is the #1 thing that stops odor.
3. Empty urine bucket when full. Dilute 1 part urine +10 parts water, apply to non-edible plants, trees, ornamental shrubs. *Do not use on salad greens, carrots or other raw food crops.*
4. When solids bucket is ~¾ full:
   - Lift it out, replace with empty second bucket.
   - Move full bucket to a covered compost curing bin (separate outdoor bin). Keep it covered, aerate by turning occasionally.
   - Let cure: **minimum 12 months cold compost; 3–6 months hot compost if pile maintains >55°C**.
   - After curing, the material should look like dark earth with no visible waste.

## Maintenance & common problems ##

- Smell: almost always from too little carbon, wet solids, or urine leaking into solids. Add more dry sawdust.
- Flies: cover solids well, keep bucket lidded when not in use, add vent pipe.
- Bucket mold: normal; scrub buckets between batches.
- Never add: plastic, wipes, chemicals, meat, oil. Only human waste + dry carbon material.

**Simple low-budget alternative (no urine diversion, caveat)**

You can build a simpler single-bucket compost toilet without urine separation, but it will smell much worse and compost slower. You have to use way more sawdust every time to soak up urine. Urine diversion is strongly recommended.

---

**Critical Health Warning**

Even after composting, human waste can carry pathogens. Do **not** apply finished compost to vegetable gardens where produce is eaten raw. Many local health departments prohibit using humanure on food crops entirely.
